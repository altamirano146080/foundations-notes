# 11_01 Attention mechanisms

In this notebook you will implement the SelfAttention class from scratch in PyTorch, translating the mathematical formula of scaled dot-product self-attention into working code. Your task is to build the `__init__` method to define the learnable weight matrices for queries, keys, and values, and then complete the `forward` method to compute attention scores, apply scaling and softmax, and return the weighted combination of values.

**Note:** The class must be able to handle a whole batch of data.

To check your implementation, you can do the following:

* Run a forward pass on random data.

* Check whether output shapes are correct.

* Verify that each row of the attention weights sums (approximately) to 1.


```python
import math
import torch
import torch.nn as nn


class SelfAttention(nn.Module):

    def __init__(self, d_model, d_k, d_v):
        super(SelfAttention, self).__init__()

        self.W_Q = nn.Linear(d_model, d_k, bias =False)
        self.W_K = nn.Linear(d_model, d_k, bias =False)
        self.W_V = nn.Linear(d_model, d_v, bias =False)

        self.scale = math.sqrt(d_k)

    def forward(self, X):
        Q = self.W_Q(X) 
        K = self.W_K(X)  
        V = self.W_V(X)  

        scores = torch.matmul(Q, K.transpose(-2, -1)) / self.scale

        attention_weights = torch.softmax(scores, dim=-1)

        outputs = torch.matmul(attention_weights, V)

        return outputs
```
